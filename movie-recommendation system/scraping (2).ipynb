```python
# Cell A: generate genre_text (prefer 'genre'), mood_tags, top_keywords, and final_text
import ast, numpy as np, pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

DATA_PATH = "Top_movies.csv"
OUT_PATH = "Top_movies_features.csv"

df = pd.read_csv(DATA_PATH)

# ensure overview exists
df['overview'] = df['overview'].fillna('').astype(str)

# --- helper: normalize existing genre column to a clean space-separated string ---
def fix_genre_text_from_column(x):
    if pd.isna(x):
        return ""
    if isinstance(x, list):
        return " ".join([str(i) for i in x])
    s = str(x).strip()
    # try parse list-like strings like "['Drama','Crime']"
    try:
        val = ast.literal_eval(s)
        if isinstance(val, list):
            return " ".join(val)
    except Exception:
        # replace common delimiters with space
        s = s.replace("|"," ").replace(",", " ").replace(";", " ")
        return " ".join([t for t in s.split() if t.strip()])
    return s

# build genre_text using the 'genre' column when available
if 'genre' in df.columns:
    df['genre_text'] = df['genre'].apply(fix_genre_text_from_column).str.strip()
else:
    df['genre_text'] = ""

# --- parse genre_ids to list (if present) ---
def parse_genre_ids(x):
    if pd.isna(x): return []
    if isinstance(x, list): return x
    if isinstance(x, (int, float)): return [int(x)]
    s = str(x)
    try:
        val = ast.literal_eval(s)
        if isinstance(val, list):
            return [int(i) for i in val]
    except Exception:
        parts = [p.strip() for p in s.replace(",", "|").split("|") if p.strip()]
        out = []
        for p in parts:
            try:
                out.append(int(p))
            except:
                pass
        return out
    return []

if 'genre_ids' in df.columns:
    df['genre_ids_list'] = df['genre_ids'].apply(parse_genre_ids)
else:
    df['genre_ids_list'] = [[] for _ in range(len(df))]

# --- TMDB mapping (fallback) ---
TMDB_GENRE_MAP = {
    28: 'Action',12:'Adventure',16:'Animation',35:'Comedy',80:'Crime',99:'Documentary',18:'Drama',
    10751:'Family',14:'Fantasy',36:'History',27:'Horror',10402:'Music',9648:'Mystery',10749:'Romance',
    878:'Science_Fiction',10770:'TV_Movie',53:'Thriller',10752:'War',37:'Western'
}

def genre_text_from_ids(ids):
    return " ".join([TMDB_GENRE_MAP.get(i, f"genre_{i}") for i in ids])

# Fill missing genre_text from genre_ids_list
mask_empty = df['genre_text'].fillna("").str.strip() == ""
df.loc[mask_empty, 'genre_text'] = df.loc[mask_empty, 'genre_ids_list'].apply(genre_text_from_ids)

# final safety: replace multiple spaces, trim
df['genre_text'] = df['genre_text'].fillna("").astype(str).replace(r'\s+', ' ', regex=True).str.strip()

# ---------------------------
# Mood mapping (uses genre_text + overview)
# ---------------------------
MOOD_MAP = {
    'drama': ['emotional','serious','thoughtful'],
    'comedy': ['happy','funny','light-hearted'],
    'romance': ['romantic','feel-good','tender'],
    'horror': ['scary','dark','tense'],
    'thriller': ['suspenseful','intense','tense'],
    'action': ['exciting','adrenaline','fast-paced'],
    'crime': ['gritty','dark','suspenseful'],
    'fantasy': ['fantastical','wonder','imaginative'],
    'science_fiction': ['futuristic','speculative','mind-bending'],
    'animation': ['family','wholesome','colorful'],
    'history': ['period','historical','dramatic'],
    'mystery': ['mysterious','puzzling','intriguing'],
    'documentary': ['informative','real','educational'],
    'war': ['intense','dramatic','gritty'],
    'music': ['musical','rhythmic','uplifting'],
    'family': ['family','kid-friendly','wholesome'],
    'western': ['western','stoic','epic']
}

# create reverse mapping of tokens -> mood tags
token_to_moods = {}
for g, moods in MOOD_MAP.items():
    token = g.replace('_', ' ').lower()
    token_to_moods[token] = moods
    token_to_moods[g.lower()] = moods

# simple keyword heuristics for overview
overview_kw_map = {
    ('love','heart','romanc','relationship','affair'): ['romantic','feel-good'],
    ('war','battle','soldier','military'): ['intense','dramatic'],
    ('prison','escape','prisoner','cell'): ['intense','gritty'],
    ('laugh','funny','humor','comedy'): ['happy','funny'],
    ('murder','kill','detective','crime','police'): ['suspenseful','gritty'],
    ('ghost','haunt','scare','horror'): ['scary','tense'],
}

def generate_mood_tags(genre_text, overview_text):
    tags = set()
    txt = (genre_text + " " + overview_text).lower()
    # genre token matching
    for token, moods in token_to_moods.items():
        if token in txt:
            tags.update(moods)
    # overview keyword heuristics
    for keys, moods in overview_kw_map.items():
        for k in keys:
            if k in txt:
                tags.update(moods)
                break
    return " ".join(sorted(tags))

df['mood_tags'] = df.apply(lambda r: generate_mood_tags(r['genre_text'], r['overview']), axis=1)

# ---------------------------
# top_keywords via TF-IDF (explainability)
# ---------------------------
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
corpus = (df['overview'].fillna("") + " " + df['genre_text'].fillna("")).values
tfidf_matrix = tfidf.fit_transform(corpus)
feature_names = np.array(tfidf.get_feature_names_out())

def top_keywords_from_tfidf(row_idx, top_n=6):
    row = tfidf_matrix[row_idx]
    vals = row.toarray().ravel()
    if vals.sum() == 0:
        return []
    top_idx = np.argsort(-vals)[:top_n]
    kws = feature_names[top_idx]
    # filter zeros (rare)
    kws = [k for k, idx in zip(kws, top_idx) if vals[idx] > 0]
    return list(kws)

df['top_keywords'] = df.index.to_series().apply(lambda i: " ".join(top_keywords_from_tfidf(i, top_n=6)))

# ---------------------------
# final_text: overview + genre_text + mood_tags + top_keywords
# ---------------------------
df['final_text'] = (
    df['overview'].fillna("").astype(str) + " " +
    df['genre_text'].fillna("").astype(str) + " " +
    df['mood_tags'].fillna("").astype(str) + " " +
    df['top_keywords'].fillna("").astype(str)
).str.replace(r'\s+', ' ', regex=True).str.strip()

# Save results
df.to_csv(OUT_PATH, index=False)
print("Option B features saved to:", OUT_PATH)
print(df[['id','title','genre_text','mood_tags','top_keywords']].head().to_string(index=False))
```

```python
# Cell A2: one-hot genres + numeric features (popularity, vote_average, vote_count, release_year)
import pandas as pd
import numpy as np
import os
from sklearn.preprocessing import MinMaxScaler

IN_PATH = "Top_movies_features.csv"   # file you produced in Step 1
OUT_PATH = "Top_movies2_features.csv"

df = pd.read_csv(IN_PATH)

# 1) Ensure columns exist
if 'genre_text' not in df.columns:
    raise KeyError("'genre_text' not found in CSV. Run Step 1 first.")

# 2) One-hot encode genres (split tokens by space)
# Build a set of genre tokens across dataset
def split_genres(s):
    if pd.isna(s) or str(s).strip()=="":
        return []
    return [t.strip().replace(" ", "_") for t in str(s).split() if t.strip()]

df['genre_tokens'] = df['genre_text'].apply(split_genres)
all_genres = sorted({g for tokens in df['genre_tokens'] for g in tokens})

# create one-hot columns
for g in all_genres:
    col = f"genre__{g}"
    df[col] = df['genre_tokens'].apply(lambda toks: 1 if g in toks else 0)

print(f"Created {len(all_genres)} genre one-hot columns: {all_genres[:10]}...")

# 3) Extract release_year from release_date if present
def extract_year(x):
    try:
        s = str(x)
        if s.strip()=="" or s.lower().strip()=="nan":
            return np.nan
        # try common formats: YYYY-MM-DD
        return int(s[:4])
    except:
        return np.nan

df['release_year'] = df['release_date'].apply(extract_year) if 'release_date' in df.columns else np.nan

# 4) Numeric features to use: popularity, vote_average, vote_count, recency (derived)
numeric_cols = []
for c in ['popularity','vote_average','vote_count']:
    if c in df.columns:
        numeric_cols.append(c)
    else:
        # create default if missing
        df[c] = 0.0
        numeric_cols.append(c)

# recency = current_year - release_year (smaller => more recent). we'll invert so larger => more recent
import datetime
current_year = datetime.datetime.now().year
df['release_year_filled'] = df['release_year'].fillna(current_year)  # missing treated as current to avoid big recency
df['recency'] = (current_year - df['release_year_filled']).clip(lower=0)
# convert to "recentness" (larger => more recent)
df['recentness'] = df['recency'].max() - df['recency']

# include recentness in numeric features
numeric_cols.append('recentness')

# 5) Scale numeric features to [0,1] using MinMaxScaler
scaler = MinMaxScaler()
numeric_matrix = scaler.fit_transform(df[numeric_cols].fillna(0).values.astype(float))
numeric_feature_names = [f"num__{c}" for c in numeric_cols]
for i, name in enumerate(numeric_feature_names):
    df[name] = numeric_matrix[:, i]

print("Numeric columns scaled and added:", numeric_feature_names)

# 6) Save resulting dataframe
df.to_csv(OUT_PATH, index=False)
print("Step 2 feature file saved to:", OUT_PATH)
print("Sample of new columns (first 5 rows):")
df[['id','title','genre_text','mood_tags'] + numeric_feature_names + [f"genre__{g}" for g in all_genres[:6]]].head()
```

```python
# Cell B2: build hybrid embeddings (textual embeddings concatenated with numeric one-hot features) and index with FAISS
import numpy as np, pandas as pd, faiss, os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

IN_PATH = "Top_movies2_features.csv"
OUT_INDEX = "faiss_hybrid.index"
OUT_EMB = "faiss_hybrid_embeddings.npy"
OUT_META = "Top_movies2_features_plus_index.csv"

df = pd.read_csv(IN_PATH)

# 1) Build textual embeddings (sentence-transformer preferred, else TF-IDF+SVD fallback)
use_st = False
try:
    from sentence_transformers import SentenceTransformer
    model = SentenceTransformer('all-MiniLM-L6-v2')
    texts = df['final_text'].fillna("").tolist()
    text_embeddings = model.encode(texts, show_progress_bar=True, convert_to_numpy=True).astype('float32')
    print("Used SentenceTransformer. dim:", text_embeddings.shape[1])
    use_st = True
except Exception as e:
    print("SentenceTransformer not available or failed:", e)
    # fallback to TF-IDF + SVD
    tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
    X = tfidf.fit_transform(df['final_text'].fillna("").values)
    svd = TruncatedSVD(n_components=min(384, X.shape[1]-1), random_state=42)
    text_embeddings = svd.fit_transform(X).astype('float32')
    print("Used TF-IDF+SVD embedding. dim:", text_embeddings.shape[1])

# 2) Build numeric vector for each movie
# choose numeric columns saved in Step2 (prefix 'num__') and genre one-hot columns 'genre__'
num_cols = [c for c in df.columns if c.startswith('num__')]
genre_onehot_cols = [c for c in df.columns if c.startswith('genre__')]

print("Numeric columns:", num_cols)
print("Genre one-hot columns (count):", len(genre_onehot_cols))

numeric_matrix = df[num_cols + genre_onehot_cols].fillna(0).values.astype('float32')

# 3) Scale numeric part to have roughly comparable magnitude to text embeddings
# We'll L2-normalize each part separately and then scale numeric part by a weight factor alpha.
from sklearn.preprocessing import normalize
text_norm = normalize(text_embeddings, axis=1)  # unit-length
num_norm = normalize(numeric_matrix, axis=1) if numeric_matrix.shape[1] > 0 else np.zeros((text_norm.shape[0],0), dtype='float32')

# alpha controls importance of numeric features relative to text (tuneable)
alpha = 0.4   # 0.0 => ignore numeric; 1.0 => numeric same weight as text
if num_norm.shape[1] > 0:
    hybrid = np.hstack([text_norm, alpha * num_norm]).astype('float32')
else:
    hybrid = text_norm.astype('float32')

# 4) Final normalization (so IndexFlatIP ~ cosine)
faiss.normalize_L2(hybrid)

# 5) Build FAISS index
d = hybrid.shape[1]
index = faiss.IndexFlatIP(d)
index.add(hybrid)
print("Built FAISS hybrid index with", index.ntotal, "vectors, dim=", d)

# 6) Save artifacts
np.save(OUT_EMB, hybrid)
faiss.write_index(index, OUT_INDEX)
df.to_csv(OUT_META, index=False)
print("Saved embeddings to", OUT_EMB)
print("Saved FAISS index to", OUT_INDEX)
print("Saved meta CSV to", OUT_META)

# 7) Quick test: top-5 for index 0
D,I = index.search(hybrid[0:1], 6)
print("Top indices for item 0:", I[0].tolist())
for idx, score in zip(I[0].tolist(), D[0].tolist()):
    if idx==0: continue
    print(df.loc[idx, ['id','title','genre_text','mood_tags']].to_dict(), "score=", float(score))
```

```python
# B2_fast.py — TF-IDF + SVD fallback only (fast)
import time
import numpy as np, pandas as pd, faiss
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import normalize

IN_PATH = "Top_movies2_features1.csv"
OUT_INDEX = "faiss_hybrid_fast.index"
OUT_EMB = "faiss_hybrid_fast_embeddings.npy"
OUT_META = "Top_movies2_features1_plus_index.csv"

t0 = time.time()
df = pd.read_csv(IN_PATH)
texts = df['final_text'].fillna("").tolist()
print("Rows:", len(texts))

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
X = tfidf.fit_transform(texts)
print("TF-IDF shape:", X.shape)
svd_dim = min(384, X.shape[1]-1)
svd = TruncatedSVD(n_components=svd_dim, random_state=42)
embeddings = svd.fit_transform(X).astype('float32')
print("SVD done, dim:", embeddings.shape[1])

# numeric part
num_cols = [c for c in df.columns if c.startswith("num__")]
genre_cols = [c for c in df.columns if c.startswith("genre__")]
numeric_matrix = df[num_cols + genre_cols].fillna(0).values.astype('float32') if (len(num_cols)+len(genre_cols))>0 else np.zeros((len(df),0),dtype='float32')

text_norm = normalize(embeddings, axis=1)
num_norm = normalize(numeric_matrix, axis=1) if numeric_matrix.shape[1]>0 else np.zeros((text_norm.shape[0],0),dtype='float32')
alpha = 0.4
hybrid = np.hstack([text_norm, alpha * num_norm]).astype('float32')
faiss.normalize_L2(hybrid)

index = faiss.IndexFlatIP(hybrid.shape[1])
index.add(hybrid)
np.save(OUT_EMB, hybrid)
faiss.write_index(index, OUT_INDEX)
df.to_csv(OUT_META, index=False)
print("Built index with", index.ntotal, "rows. Time:", time.time()-t0)
```

```python
import numpy as np, faiss, pandas as pd

# paths you used
EMB_PATH = "faiss_hybrid_fast_embeddings.npy"
INDEX_PATH = "faiss_hybrid_fast.index"
META_PATH = "Top_movies2_features1_plus_index.csv"  # was saved by the script

# load
emb = np.load(EMB_PATH).astype('float32')
index = faiss.read_index(INDEX_PATH)
meta = pd.read_csv(META_PATH)

# helper maps
id_to_idx = pd.Series(meta.index.values, index=meta['id'].astype(str)).to_dict()
title_to_idx = pd.Series(meta.index.values, index=meta['title'].astype(str)).to_dict()

def recommend_hybrid_by_index(query_idx, top_k=10, filter_self=True):
    q = emb[query_idx].reshape(1, -1).astype('float32')
    D, I = index.search(q, top_k + 20)  # request extras for filtering
    results = []
    seen_ids = set()
    for score, ridx in zip(D[0].tolist(), I[0].tolist()):
        if filter_self and ridx == query_idx:
            continue
        movie_id = str(meta.loc[ridx,'id'])
        if movie_id in seen_ids:
            continue
        seen_ids.add(movie_id)
        results.append({
            'row_index': int(ridx),
            'movie_id': int(meta.loc[ridx,'id']),
            'title': meta.loc[ridx,'title'],
            'genres': meta.loc[ridx,'genre_text'],
            'mood_tags': meta.loc[ridx,'mood_tags'] if 'mood_tags' in meta.columns else "",
            'top_keywords': meta.loc[ridx,'top_keywords'] if 'top_keywords' in meta.columns else "",
            'score': float(score)
        })
        if len(results) >= top_k:
            break
    return pd.DataFrame(results)

def recommend_hybrid_by_movie_id(movie_id, top_k=10):
    movie_id = str(movie_id)
    if movie_id not in id_to_idx:
        raise KeyError(f"movie id {movie_id} not found")
    return recommend_hybrid_by_index(id_to_idx[movie_id], top_k=top_k)

def recommend_hybrid_by_title(title, top_k=10):
    if title not in title_to_idx:
        raise KeyError(f"title '{title}' not found")
    return recommend_hybrid_by_index(title_to_idx[title], top_k=top_k)

# Example usage:
print("Query example (row 0):", meta.loc[0, ['id','title','genre_text']].to_dict())
print(recommend_hybrid_by_index(0, top_k=6))
# or by id/title:
# print(recommend_hybrid_by_movie_id(278, top_k=6))
# print(recommend_hybrid_by_title("The Godfather", top_k=6))
```

```python
# 1) Quick check: do the files exist?
import os
print("Exists meta:", os.path.exists("Top_movies2_features1_plus_index.csv"))
print("Exists embeddings:", os.path.exists("faiss_hybrid_fast_embeddings.npy"))
print("Exists index:", os.path.exists("faiss_hybrid_fast.index"))

# 2) Improved recommend_by_mood_hybrid with progress and safe fallbacks
def recommend_by_mood_hybrid(mood_str, top_k=20, use_faiss=True, rerank_by='score', debug=False):
    import numpy as np, pandas as pd, faiss, time
    t0 = time.time()
    meta = pd.read_csv("Top_movies2_features1_plus_index.csv")
    mood_str = str(mood_str).strip().lower()
    if mood_str == "":
        if debug: print("Empty mood string -> returning empty DataFrame")
        return pd.DataFrame([])

    # mask (match whole or partial tokens)
    mask = meta['mood_tags'].fillna("").str.lower().str.contains(mood_str)
    candidates = meta[mask].copy()
    if candidates.shape[0] == 0:
        if debug: print("No candidates found for mood:", mood_str)
        return pd.DataFrame([])

    if not use_faiss:
        col = rerank_by if rerank_by in candidates.columns else 'vote_average'
        candidates[col] = pd.to_numeric(candidates[col], errors='coerce').fillna(0)
        out = candidates.sort_values(by=col, ascending=False).head(top_k)[['id','title','genre_text','mood_tags','vote_average','popularity']]
        if debug: print(f"Returned {len(out)} rows (no FAISS). Time {time.time()-t0:.2f}s")
        return out

    # use FAISS path
    if debug: print("Loading embeddings and index...", flush=True)
    emb_path = "faiss_hybrid_fast_embeddings.npy"
    idx_path = "faiss_hybrid_fast.index"
    if not os.path.exists(emb_path) or not os.path.exists(idx_path):
        raise FileNotFoundError("Embeddings or index not found. Check paths.")
    emb = np.load(emb_path).astype('float32')
    # note: we don't need to load the global index here; we'll build a small index for candidates
    candidate_rows = candidates.index.tolist()
    if debug: print(f"Found {len(candidate_rows)} candidate rows for mood '{mood_str}'", flush=True)

    # if very few candidates, just sort by rerank column
    if len(candidate_rows) <= top_k:
        if debug: print("Few candidates <= top_k, sorting by rerank or score", flush=True)
        if rerank_by in candidates.columns:
            candidates[rerank_by] = pd.to_numeric(candidates[rerank_by], errors='coerce').fillna(0)
            out = candidates.sort_values(by=rerank_by, ascending=False).head(top_k)[['id','title','genre_text','mood_tags','vote_average','popularity']]
            if debug: print(f"Returned {len(out)} rows. Time {time.time()-t0:.2f}s")
            return out
        else:
            out = candidates.head(top_k)[['id','title','genre_text','mood_tags','vote_average','popularity']]
            if debug: print(f"Returned {len(out)} rows. Time {time.time()-t0:.2f}s")
            return out

    # Build candidate embeddings and run FAISS search using a mood "seed" (mean vector)
    cand_embs = emb[candidate_rows]
    # seed vector = mean of candidate embeddings (represents the mood centroid), normalized
    seed = cand_embs.mean(axis=0, keepdims=True).astype('float32')
    faiss.normalize_L2(seed)
    faiss.normalize_L2(cand_embs)
    # small index for candidate embeddings
    cand_index = faiss.IndexFlatIP(cand_embs.shape[1])
    cand_index.add(cand_embs)
    k = min(top_k, cand_embs.shape[0])
    D, I = cand_index.search(seed, k)
    picked = [candidate_rows[i] for i in I[0].tolist()]
    res = meta.loc[picked][['id','title','genre_text','mood_tags','vote_average','popularity']].copy()
    res['score'] = D[0].tolist()
    res = res.sort_values(by='score', ascending=False).head(top_k)
    if debug: print(f"FAISS mood-hybrid returned {len(res)} rows. Time {time.time()-t0:.2f}s")
    return res.reset_index(drop=True)

# 3) Call examples (run these to see output)
print("\nExample: mood='emotional' (FAISS path):")
print(recommend_by_mood_hybrid('emotional', top_k=10, use_faiss=True, debug=True).head(10).to_string(index=False))

print("\nExample: mood='romantic' (non-FAISS quick path):")
print(recommend_by_mood_hybrid('romantic', top_k=10, use_faiss=False, debug=True).head(10).to_string(index=False))
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import normalize
import numpy as np
import faiss


META_PATH = "Top_movies2_features1_plus_index.csv"
meta = pd.read_csv(META_PATH)

use_st = False

try:
    from sentence_transformers import SentenceTransformer
    st_model = SentenceTransformer("all-MiniLM-L6-v2")
    use_st = True
    print("SentenceTransformer loaded")
except:
    print("Using TFIDF+SVD")


def build_hybrid_query_embedding(text):

    text = str(text)

    # TEXT
    if use_st:
        text_emb = st_model.encode([text]).astype("float32")
    else:
        tfidf = TfidfVectorizer(max_features=5000, stop_words="english")
        X = tfidf.fit_transform(meta["final_text"].fillna(""))
        svd = TruncatedSVD(n_components=384, random_state=42)
        svd.fit(X)

        q = tfidf.transform([text])
        text_emb = svd.transform(q).astype("float32")

    text_norm = normalize(text_emb, axis=1)

    # numeric part
    num_cols = [c for c in meta.columns if c.startswith("num__")]
    genre_cols = [c for c in meta.columns if c.startswith("genre__")]

    dim = len(num_cols) + len(genre_cols)

    if dim > 0:
        num_vec = np.zeros((1, dim), dtype="float32")
        num_norm = normalize(num_vec, axis=1)
        alpha = 0.4
        hybrid = np.hstack([text_norm, alpha * num_norm])
    else:
        hybrid = text_norm

    faiss.normalize_L2(hybrid)

    return hybrid.astype("float32")
```

```python
def suggest(query, by='title', top_k=6):

    by = by.lower()

    # ---------- TITLE ----------
    if by == 'title':
        q_idx = meta[meta['title'].str.lower() == query.lower()].index[0]
        q_emb = emb[q_idx: q_idx+1]

    # ---------- FREE TEXT ----------
    elif by == 'text':
        # FIX: must use hybrid embedding
        q_emb = build_hybrid_query_embedding(query)

    # ---------- MOOD ----------
    elif by == 'mood':
        return recommend_by_mood_fixed(query, top_k)

    else:
        raise ValueError("by must be 'title', 'text', or 'mood'")

    # ---------- FAISS SEARCH ----------
    faiss.normalize_L2(q_emb)
    D, I = index.search(q_emb.astype('float32'), top_k)

    res = meta.iloc[I[0]].reset_index().rename(columns={'index':'row_index'})
    res['score'] = D[0]

    return _attach_explanations_for_query_fixed(
        q_idx if by == 'title' else None,
        res[['row_index', 'score']]
    )
```

```python
# Fix: recommend_by_mood + helpers that always include row_index, then run examples
import os, numpy as np, pandas as pd, faiss

META_PATH = "Top_movies2_features1_plus_index.csv"
EMB_PATH = "faiss_hybrid_fast_embeddings.npy"
INDEX_PATH = "faiss_hybrid_fast.index"

meta = pd.read_csv(META_PATH)
emb = np.load(EMB_PATH).astype('float32')
index = faiss.read_index(INDEX_PATH)

# Re-define _attach_explanations_for_query but expecting df_res includes 'row_index'
def _attach_explanations_for_query_fixed(query_idx, df_res):
    q_genres = set()
    q_moods = set()
    q_kws = set()
    if query_idx is not None:
        q_genres = set(str(meta.loc[query_idx,'genre_text']).lower().split())
        q_moods = set(str(meta.loc[query_idx,'mood_tags']).lower().split()) if 'mood_tags' in meta.columns else set()
        q_kws = set(str(meta.loc[query_idx,'top_keywords']).split()) if 'top_keywords' in meta.columns else set()
    # if no explicit query index (like free-text or mood), build proxy from first returned item if present
    if query_idx is None and len(df_res) > 0:
        proxy_idx = int(df_res.iloc[0]['row_index'])
        q_genres = set(str(meta.loc[proxy_idx,'genre_text']).lower().split())
        q_moods = set(str(meta.loc[proxy_idx,'mood_tags']).lower().split()) if 'mood_tags' in meta.columns else set()
        q_kws = set(str(meta.loc[proxy_idx,'top_keywords']).split()) if 'top_keywords' in meta.columns else set()

    expl_rows = []
    for _, r in df_res.iterrows():
        ridx = int(r['row_index'])
        cand_genres = set(str(meta.loc[ridx,'genre_text']).lower().split())
        cand_moods = set(str(meta.loc[ridx,'mood_tags']).lower().split()) if 'mood_tags' in meta.columns else set()
        cand_kws = set(str(meta.loc[ridx,'top_keywords']).split()) if 'top_keywords' in meta.columns else set()
        genre_overlap = sorted(list(q_genres & cand_genres))
        mood_overlap = sorted(list(q_moods & cand_moods))
        shared_keywords = sorted(list(q_kws & cand_kws))
        why = []
        if genre_overlap:
            why.append("Genres: " + ", ".join(genre_overlap))
        if mood_overlap:
            why.append("Mood: " + ", ".join(mood_overlap))
        if shared_keywords:
            why.append("Keywords: " + ", ".join(shared_keywords[:3]))
        why_text = " — ".join(why) if why else "Similar by semantic features."
        expl_rows.append({
            'row_index': ridx,
            'movie_id': int(meta.loc[ridx,'id']),
            'title': meta.loc[ridx,'title'],
            'genres': meta.loc[ridx,'genre_text'],
            'mood_tags': meta.loc[ridx,'mood_tags'] if 'mood_tags' in meta.columns else "",
            'top_keywords': meta.loc[ridx,'top_keywords'] if 'top_keywords' in meta.columns else "",
            'score': float(r.get('score', 0.0)),
            'genre_overlap': genre_overlap,
            'mood_overlap': mood_overlap,
            'shared_keywords': shared_keywords,
            'why': why_text
        })
    return pd.DataFrame(expl_rows)

# Fixed recommend_by_mood that always returns row_index and uses the fixed expl function
def recommend_by_mood_fixed(mood_str, top_k=20, use_faiss=True, rerank_by='score', debug=False):
    mood = str(mood_str).strip().lower()
    cand_mask = meta['mood_tags'].fillna("").str.lower().str.contains(mood)
    candidates = meta[cand_mask].copy()
    if candidates.shape[0] == 0:
        if debug: print("No mood candidates found for:", mood)
        return pd.DataFrame([])

    # preserve the original row indices for lookups
    candidates = candidates.reset_index().rename(columns={'index':'row_index'})

    if not use_faiss:
        col = rerank_by if rerank_by in candidates.columns else 'vote_average'
        if col in candidates.columns:
            candidates[col] = pd.to_numeric(candidates[col], errors='coerce').fillna(0)
            out = candidates.sort_values(by=col, ascending=False).head(top_k)
        else:
            out = candidates.head(top_k)
        # ensure returned df has row_index
        return _attach_explanations_for_query_fixed(None, out[['row_index'] + out.columns.drop('row_index').tolist()])

    # FAISS path: build candidate embeddings using row_index
    candidate_rows = candidates['row_index'].tolist()
    cand_embs = emb[candidate_rows]
    # if very few candidates, fallback to simple rerank
    if len(candidate_rows) <= top_k:
        if rerank_by in candidates.columns:
            candidates[rerank_by] = pd.to_numeric(candidates[rerank_by], errors='coerce').fillna(0)
            out = candidates.sort_values(by=rerank_by, ascending=False).head(top_k)
            return _attach_explanations_for_query_fixed(None, out[['row_index'] + out.columns.drop('row_index').tolist()])
        else:
            out = candidates.head(top_k)
            return _attach_explanations_for_query_fixed(None, out[['row_index'] + out.columns.drop('row_index').tolist()])

    # build small index and search by centroid
    seed = cand_embs.mean(axis=0, keepdims=True).astype('float32')
    faiss.normalize_L2(seed)
    faiss.normalize_L2(cand_embs)
    cand_index = faiss.IndexFlatIP(cand_embs.shape[1])
    cand_index.add(cand_embs)
    k = min(top_k, cand_embs.shape[0])
    D, I = cand_index.search(seed, k)
    picked = [candidate_rows[i] for i in I[0].tolist()]
    res = meta.loc[picked].reset_index().rename(columns={'index':'row_index'})  # attach row_index
    res = res[['row_index'] + [c for c in res.columns if c != 'row_index']]
    res['score'] = D[0].tolist()
    return _attach_explanations_for_query_fixed(None, res)

# Replace the old functions in your notebook namespace
recommend_by_mood = recommend_by_mood_fixed
_attach_explanations_for_query = _attach_explanations_for_query_fixed

# ---------- Run the three examples again so you see consistent outputs ----------
print("By title example (The Shawshank Redemption):")
display(suggest("The Shawshank Redemption", by='title', top_k=6))

print("\nBy free text example ('heart-warming romantic drama'):")
display(suggest("heart-warming romantic drama", by='text', top_k=6))

print("\nBy mood example ('emotional'):")
display(suggest("emotional", by='mood', top_k=6)) 
```

```python
ott_cache = {}

def add_ott_to_result(df):

    df = df.copy()

    if "movie_id" not in df.columns:
        print("movie_id missing")
        return df

    df["ott"] = ""
    df["ott_link"] = ""

    for idx, movie_id in zip(df.index, df["movie_id"]):

        if movie_id in ott_cache:
            ott, link = ott_cache[movie_id]
        else:
            ott, link = get_ott(movie_id)
            ott_cache[movie_id] = (ott, link)

        df.loc[idx, "ott"] = ott
        df.loc[idx, "ott_link"] = link

    return df
```

```python
def apply_preferences(
    df,
    genre=None,
    mood=None,
    min_rating=None,
    min_year=None
):

    df = df.copy()

    # genre filter
    if genre is not None and "genres" in df.columns:
        df = df[df["genres"].str.contains(genre, case=False, na=False)]

    # mood filter
    if mood is not None and "mood_tags" in df.columns:
        df = df[df["mood_tags"].str.contains(mood, case=False, na=False)]

    # rating filter
    if min_rating is not None and "vote_average" in df.columns:
        df = df[df["vote_average"] >= min_rating]

    # year filter (from release_date)
    if min_year is not None and "release_date" in df.columns:

        df["year_tmp"] = df["release_date"].astype(str).str[:4]
        df["year_tmp"] = pd.to_numeric(df["year_tmp"], errors="coerce")

        df = df[df["year_tmp"] >= min_year]

    return df
```

```python
def recommend_with_preferences(
    title=None,
    text=None,
    mood=None,
    genre=None,
    min_rating=None,
    min_year=None,
    top_k=5
):

    # ---------- base ----------
    if title is not None:
        res = suggest(title, by="title", top_k=20)

    elif text is not None:
        res = suggest(text, by="text", top_k=20)

    elif mood is not None:
        res = suggest(mood, by="mood", top_k=20)

    else:
        res = meta.copy()

        # FIX → id to movie_id
        if "movie_id" not in res.columns and "id" in res.columns:
            res = res.rename(columns={"id": "movie_id"})

        if "why" not in res.columns:
            res["why"] = "Filtered by preferences"


    # ---------- filter ----------
    res = apply_preferences(
        res,
        genre=genre,
        mood=mood,
        min_rating=min_rating,
        min_year=min_year
    )

    res = res.head(top_k)

    # ---------- add ott ----------
    res = add_ott_to_result(res)

    if "ott" not in res.columns:
        res["ott"] = ""

    if "ott_link" not in res.columns:
        res["ott_link"] = ""

    return res
```
